```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# from google.colab import data_table
# data_table.enable_dataframe_formatter()

print("Libraries loaded as expected")
```

```python
df = pd.read_csv('Planetary_data.csv', comment='#')
```

```python
# The question we're trying to answer are heavy planets always wide.

subset_data = df.dropna(subset=['pl_bmasse', 'pl_rade']).copy()
```

```python
subset_data.plot(kind='scatter', x='pl_bmasse', y='pl_rade', xlabel="Planet mass", ylabel='planet radius')
```

```python
subset_data2 = (df.dropna(subset=['pl_bmasse', 'pl_rade']).query('pl_bmasse < 1000').copy())

subset_data2.plot(kind="scatter", x='pl_bmasse', y='pl_rade', xlabel='Planet mass', ylabel='Planet radius')
```
