```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# from google.colab import data_table
# data_table.enable_dataframe_formatter()

print("Libraries loaded as expected")
```

```python
df = pd.read_csv('Planetary_data.csv', comment='#')
```

```python
kepler_data = df['pl_name'].str.contains('Kepler')
print(kepler_data.value_counts())
```

```python
conditions = [
    (df['pl_name'].str.contains('Kepler')),
    (df['pl_name'].str.contains('TOI')),
    (df['pl_name'].str.contains('WASP')),
]

choices = ['Kepler','TESS', 'WASP']

df['filtered_search_data'] = np.select(condlist=conditions, choicelist=choices, default='Other')
print(df['filtered_search_data'].value_counts())
```

```python
data_to_be_plotted = df['filtered_search_data'].value_counts()
data_to_be_plotted.plot(kind='bar', xlabel='Survey methods to find exoplanets'  , title='Survey methods distribution')
```
